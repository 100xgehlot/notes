```python
def generate_e2():
    return e1**d%p

def generate_c1():
    return e1**r%p

def generate_c2():
    return (pt*e2**r)%p

def encryption():
    return (c1,c2)

def decryption():
    temp = pow(c1,d,p)

    temp_inv = pow(temp,-1,p)
    dpt = (c2*temp_inv)%p
    return dpt
p = int(input("Enter 1st part of public key(p): "))
e1 = int(input("Enter 2nd part of public key(e1):"))
d = int(input("Enter private key (d):"))
r = int(input("Enter random integer key(r): "))
pt = int(input("Enter the plain text(pt): "))
e2  = generate_e2()
c1 = generate_c1()
c2 = generate_c2()
print("Encrypyed text: ",encryption())
print("decrypted text: ",decryption())
```
