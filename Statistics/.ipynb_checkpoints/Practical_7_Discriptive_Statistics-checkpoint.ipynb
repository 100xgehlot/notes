```python
import pandas as pd
df = pd.read_csv("stats.csv")
df
```

```python
#Measure of Central Tendancy
#Mean Salary
mean1=df["Salary"].mean()
mean1
```

```python
#Sum of Salaries
sum1 = df["Salary"].sum()
sum1
```

```python
#Max Salary
max1 = df["Salary"].max()
max1
```

```python
#Min Salary
min1 = df["Salary"].min()
min1
```

```python
#Total Count( of row)
count1 = df["Salary"].count()
count1
```

```python
#Median of Salary
median1 = df["Salary"].median()
median1
```

```python
#Mode of Salary
mode1 = df["Salary"].mode()
mode1
```

```python
country_wise_sum = df.groupby(["Country"])["Salary"].sum()
country_wise_sum
```

```python
country_wise_count = df.groupby(["Country"]).count()
country_wise_count
```

```python
#Measure of Variability
var1 = df["Salary"].var()
var1
```

```python
std1 = df["Salary"].std()
std1
```

```python
#Measure of Symmetry
skew1 = df.skew(axis=0,skipna=True)
skew1
```

```python
#The skewness is positive so x will have right side tail.
df.describe()
```
