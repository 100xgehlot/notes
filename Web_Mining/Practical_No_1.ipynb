```python
from bs4 import BeautifulSoup
import requests

url = "http://olympus.realpython.org/profiles/dionysus"
response = requests.get(url)

print(response.status_code)
# Output: 200

# Extracts the HTML content from the response
soup = BeautifulSoup(response.text, "html.parser")

# Formats and prints the HTML nicely
print(soup.prettify())
```

```python
title = soup.title.string
print("Page Title:", title)
```

```python
text_content = soup.get_text()
print("Text Content:\n", text_content)
```

```python
# Find all links
links = soup.find_all("a")
for link in links:
    print("Link:", link["href"])

# Save page text to a file
with open("dionysus_profile.txt", "w", encoding="utf-8") as file:
    file.write(text_content.strip())

# Find all images
images = soup.find_all("img")

for img in images:
    img_src = img["src"]

    # Determine full image URL
    if img_src.startswith("http"):
        img_url = img_src
    else:
        base_url = "http://olympus.realpython.org"
        img_url = base_url + img_src

    # Download the image
    img_response = requests.get(img_url)
    if img_response.status_code == 200:
        img_name = img_url.split("/")[-1]
        with open(img_name, "wb") as file:
            file.write(img_response.content)
        print(f"{img_name}: Downloaded")
```
