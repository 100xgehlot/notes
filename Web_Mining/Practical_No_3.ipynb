```python
import networkx as nx
import matplotlib.pyplot as plt

# Create a directed graph
G = nx.DiGraph()

# Add edges representing links between pages
edges = [("A", "B", 1), ("B", "C", 1), ("C", "A", 1), ("C", "D", 1), ("D", "A", 1)]
G.add_weighted_edges_from(edges)
# G.add_edges_from(edges)

print(G)
```

```python
# Compute PageRank
pagerank_scores = nx.pagerank(G, alpha=0.85)  # alpha is the damping factor
print("PageRank Scores:", pagerank_scores)
```

```python
# Draw the graph with node sizes proportional to PageRank
plt.figure(figsize=(8,6))
pos = nx.spring_layout(G, seed=42)
node_sizes = [5000 * pagerank_scores[node] for node in G.nodes()]  # Scale node sizes

nx.draw(
    G, pos, with_labels=True, node_color='skyblue', edge_color='red',
    node_size=node_sizes, font_size=12, font_weight='bold'
)

plt.title("PageRank Visualization")
plt.show()
```
