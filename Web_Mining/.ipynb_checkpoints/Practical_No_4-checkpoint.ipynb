```python

# Importing necessary libraries
import pandas as pd
from mlxtend.frequent_patterns import apriori, association_rules

# Sample transaction dataset
dataset = [
    ['Milk', 'Bread', 'Butter'],
    ['Milk', 'Diaper', 'Juice', 'Bread'],
    ['Milk', 'Diaper', 'Bread'],
    ['Diaper', 'Juice', 'Bread'],
    ['Milk', 'Diaper', 'Juice', 'Bread']
]

# Extract all items
all_items = set()
for transaction in dataset:
    for item in transaction:
        if item not in all_items:
            all_items.add(item)

print("All items:", all_items)
```

```python
# One Hot Encoding the transactions and converting to Pandas DataFrame
transaction_list = []

for transaction in dataset:
    transaction_dict = {}

    for item in all_items:
        transaction_dict[item] = item in transaction

    transaction_list.append(transaction_dict)

df = pd.DataFrame(transaction_list)

df
```

```python
frequent_itemsets = apriori(df, min_support=0.6, use_colnames=True)

print("Frequent Itemsets:")
frequent_itemsets
```

```python
rules = association_rules(
    frequent_itemsets,
    metric="confidence",
    min_threshold=0.7,
    num_itemsets=2
)

print("Association Rules:")
rules
```

```python
strong_rules = rules[
    (rules['confidence'] > 0.7) & (rules['lift'] > 1)
]

print("Strong Association Rules (Confidence > 0.7 and Lift > 1):")
strong_rules.head()
```
